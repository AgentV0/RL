#1. Policy gradient
Policy gradient — это класс методов, в которых мы напрямую оптимизируем параметры θ стохастической политики πθ(a|s) для максимизации ожидаемой дисконтированной награды:$$J(\theta) = \mathbb{E}[G_0].$$

В отличие от value-based методов (Q-learning, SARSA), которые сначала оценивают Q-функцию и лишь затем извлекают из неё политику, policy gradient ищет хорошую политику без промежуточного построения точной оценки ценности.$$\nabla J = \mathbb{E} \left[ \nabla \log \pi_\theta(a|s) \cdot Q^\pi(s, a) \right]$$
Это выражение не требует дифференцируемости среды и позволяет обновлять политику градиентным восхождением.

Преимущества: естественное исследование, работа с непрерывными действиями, устойчивость к ошибкам аппроксимации.

Недостаток: высокая дисперсия градиента.

#2. REINFORCE
REINFORCE - первый и самый простой алгоритм policy gradient.Он напрямую подставляет в формулу градиента вместо Qπ(s,a) фактическую дисконтированную доходность:$$G_t = \sum_{k=0}^{\infty} \gamma^k r_{t+k}$$
Алгоритм:

*  Собрать N полных эпизодов под текущей политикой πθ
*  Для каждого перехода (st,at) вычислить Gt
*  Оценить градиент:
$$\nabla J \approx \frac{1}{N} \sum_i \sum_t \nabla \log \pi_\theta(a_t \mid s_t) \cdot G_t$$
* Обновить θ←θ+α∇J

REINFORCE является on-policy методом

Основные проблемы - высокая дисперсия (из-за случайности Gt) и необходимость дожидаться окончания эпизода. Для снижения дисперсии из Gt вычитают базовую линию b(s) (например, скользящее среднее или V(s)), что не меняет математическое ожидание градиента, но может значительно уменьшить его разброс.

#3. Actor-critic
Actor-Critic - это гибридный подход, объединяющий policy gradient (актёр) и value-based методы (критик).

Актёр πθ(a|s) учит политику

Благодаря критику можно заменить доходность Gt на преимущество A(s,a) = Q(s,a) - V(s), которое имеет гораздо меньшую дисперсию. На практике используется одношаговая или n-шаговая оценка:$$A(s, a) = r + \gamma V_\phi(s') - V_\phi(s)$$

Критик минимизирует MSE:$$\left( V_\phi(s) - \left( r + \gamma V_\phi(s') \right) \right)^2.$$

Актёр обновляется по градиенту:
$$\nabla \log \pi_\theta(a|s) \cdot A(s, a).$$

Преимущества: возможность обучения на каждом шаге, значительно меньшая дисперсия, чем в REINFORCE, и, как следствие, более стабильная сходимость. Actor-Critic лежит в основе современных алгоритмов, но его основная сложность - необходимость одновременного обучения двух сетей.
