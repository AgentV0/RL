#1. Q-learning, SARSA
**Q-learning и SARSA** - это два основных алгоритма временных разностей для оценки  Q(s,a) в model-free RL.

**Q-learning** (off-policy) использует:$$\hat{Q}(s, a) = r + \gamma \cdot \max_{a'} Q(s', a')$$

Берёт максимум по всем возможным действиям в следующем состоянии. Благодаря этому Q-learning может обучаться на данных, собранных любой политикой , и способен находить оптимальную политику π∗ независимо от того, как агент исследовал среду.

**SARSA** (on-policy) использует:$$\hat{Q}(s, a) = r + \gamma \cdot Q(s', a')$$

где
a' - это реальное действие, которое агент выбрал в состоянии s' по своей текущей политике. SARSA оценивает ценность той самой политики, которой агент действует. Он более "осторожен", например если политика включает случайные шаги, SARSA учитывает это и может избегать опасных состояний.

Ключевое различие алгоритмов в том, что Q-learning учит оптимальную политику в предположении, что в будущем агент всегда будет выбирать лучшие действия, а SARSA учит с учётом реального поведения, включая исследования.
#2. Exploration vs Exploitation
**Exploitation** - выбор действия, которое сейчас кажется наилучшим (максимизирует Q(s,a)), чтобы получить высокую награду.

**Exploration** - попытка новых или случайных действий, чтобы обнаружить потенциально более выгодные стратегии.

Если агент всегда эксплуатирует, он может навсегда застрять в субоптимуме - например, робот, научившийся ползать с положительной наградой, никогда не попробует ходить.

Основной метод баланса - ε-жадная политика:

*   с вероятностью ε выбираем случайное действие,
*   с вероятностью 1 - ε выбираем argmax_aQ(s,a).

Для сходимости к оптимальной политике ε обычно постепенно уменьшают
