# 1. V(s), Q(s, a) - что это такое?
**V(s)** - это функция ценности состояния. Она показывает ожидаемую суммарную дисконтированную доходность, которую может получить агент, начав из состояния s и далее действуя согласно некоторой политике π. Формально:$$v_\pi(s) = \mathbb{E}_\pi[G_t \mid S_t = s].$$

**Q(s, a)** - это функция ценности действия. Она показывает ожидаемую доходность, если агент находится в состоянии s, выберет действие a, а затем будет действовать по политике π:$$q_\pi(s, a) = \mathbb{E}_\pi[G_t \mid S_t = s, A_t = a].$$

Связь между ними:
Ценность состояния равна среднему по действиям от ценности действий:
$$v_{\pi}(s) = \sum_a \pi(a \mid s) \cdot q_{\pi}(s, a)$$

И наоборот:
$$q_\pi(s, a) = \sum_{r, s'} p(r, s' \mid s, a) \left[ r + \gamma v_\pi(s') \right]$$

ценность действия равна немедленному вознаграждению плюс дисконтированная ценность следующего состояния.

Эти функции лежат в основе всех алгоритмов RL.
#2. Метод Policy iteration
Policy iteration – это итеративный алгоритм для нахождения оптимальной политики в MDP. Он состоит из двух чередующихся этапов: оценка политики и ее улучшение.

**Оценка политики**

Для заданной детерминированной или стохастической политики π решается задача предсказания - вычисляется v_π(s) для всех состояний. Используется уравнение Беллмана ожидания для детерминированного случая:$$v_{\pi}(s) = \sum_{r, s'} p(r, s' \mid s, \pi(s)) [r + \gamma v_{\pi}(s')].$$ и для стохастического случая:$$v_{\pi}(s) = \sum_a \pi(a \mid s) \sum_{r, s'} p(r, s' \mid s, a) [r + \gamma v_{\pi}(s')]$$

**Улучшение политики**

Зная
v_π(s), строим новую жадную политику:$$\pi'(s) = \arg \max_a \sum_{r,s'} p(r, s' \mid s, a) [r + \gamma v_\pi(s')].$$

Жадность по отношению к v_π гарантирует, что новая политика не хуже старой, а если она не отличается, то достигнута оптимальность.

Алгоритм повторяет эти два шага до тех пор, пока политика не перестанет изменяться. На каждом шаге мы полностью оцениваем текущую политику, а затем улучшаем её. Policy iteration сходится за конечное число шагов, но каждая итерация требует вычисления v_π для всех состояний, сложность:$$O(|A||S|^2 + |S|^3)$$Однако из-за быстрой сходимости метод достаточно эффективный.
