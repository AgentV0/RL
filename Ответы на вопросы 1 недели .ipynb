#1. Базовые понятия RL
Reinforcement Learning (RL) - обучение оптимальной стратегии методом проб и ошибок. Агент взаимодействует со средой: получает состояние s, выбирает действие a, среда даёт награду r и переводит агента в новое состояние. Цель - максимизировать суммарную награду за эпизод.

**Ключевые элементы:**

*   Состояние (s) - что видит агент.
*   Действие (a) - что может сделать.
*   Награда (r) - сигнал успеха.
*   Политика π(a|s) - правило выбора действия.
*   Функция ценности V(s) - насколько хорошее данное состояние для агента с учетом возможных будущих наград
*   Марковское предположение: следующее состояние зависит только от текущего s и a.


Отличие от Supervised Learning: в SL есть готовые ответы (y), модель не влияет на данные. В RL нет "правильных" ответов, агент учится на своих действиях и их последствиях, а его выборы меняют будущие наблюдения.

# 2. Метод кросс-энтропии (табличный + нейросеть)
Cross-Entropy Method (CEM) -  играем много сессий, отбираем лучшие (элитные), переучиваем политику под их действия.

**Табличный вариант:**
*  Играем N сессий.
*  Берём M лучших по суммарной награде (элиты).
*  Для каждого s обновляем *π(a|s) = (число взятий a в элитах) / (число посещений s в элитах).*
*  Повторяем до сходимости

Работает только при малом числе состояний.
